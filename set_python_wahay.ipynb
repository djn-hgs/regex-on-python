We can find the union of two sets: $A\cup B$

This is equivalent to the `OR` binary operation and we know that `||` is the standard programming equivalent...

```python
{1,2,3} | {2, 3, 4, 5}
```

```python
{1, 2, 3}.union({2,3,4,5})
```

We can find the intersection of two sets: $A\cap B$

This is equivalent to the `AND` binary operation and we know that `&&` is the standard programming equivalent...

```python
{1,2,3} & {2, 3, 4, 5}
```

```python
{1, 2, 3}.intersection({2,3,4,5})
```

We've seen that python interprets the symbol `<` in the best way possible for the given types...

For strings it is "lexicographic" - like a dictionary.

For numbers it is numerical order.

For lists it is "lexicographic".

For sets it is "inclusion", ie subset.


```python
[1, 'A', {2,3}] < [1, 'A', 'C']
```

In maths we can form the **Cartesian product** of two sets...

$$\{1,2,3\}\times \{3,4,5\}=\{(1,3), (1,4), (1,5), (2,3), ..., (3,5)\}$$

```python
{(a, b) for a in {1,2,3} for b in {3,4,5}}
```
