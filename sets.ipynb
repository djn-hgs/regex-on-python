# What is a set?

You should remember the idea of a set from your lower school maths career!

A set is an ordered collection of objects that cannot contain repetitions and which may be infinite. Thus an array or a list is not a set, but both can be converted (irreversibly) to sets.

```python
a = [3, 2, 3, 1, 5, 2]

set(a)
```

## Set operations

The standard operations on sets $A$ and $B$ are **intersection** $A\cap B$, **union** $A\cup B$ and **difference** $A\setminus B$.

When we know the **universe** for a problem then we can also find the complement $A'$ (but the $'$ symbol can be really tricky to spot!)

There is often confusion between "universe" and "universal set". They are not the same thing (if you've a couple of hours spare then ask me, and I'll explain).

We can ask whether one set is a **subset** of another $A\subset B$ (or $A\subseteq B$).

Now let's see how well Python can handle this...

```python
a = {3, 1, 2, 5}
b = {2, 3, 1, 5, 2, 7, 9}
c = {1, 2, 3, 4, 4, 4, 5}
```

```python
c
```

```python
# We don't need funky symbols for subsets and so on because Python has a trick...

a < c
```

```python
c < a
```

```python
# You may not have realised this, but Python can do this with lots of neat types...

'apple' > 'appetite'
```

```python
'apple' < 'apology'
```

```python
[1, 2, 4] > [1, 3]
```

```python
['apple', 3, [7]] > ['apple', 3, [2, 3]]
```

## Binary operators for set operations

We've just seen that subset is just treated as the usual "less than" or "more than" operator.

We need to be a little bit more careful with some of the other symbols...

We use the `|` symbol for **union** because `||` is the standard symbol for `OR`.

We use the `&` symbol for **intersection** because `&&` is the standard symbol for `AND`.

```python
# The

a | b
```

```python
b & c
```

```python
b & c < a | b
```

This is a little tricky to read, but bear with and all will get better. Because there is no "universe" for Python sets, there is no symbol for a complement, but...

You should also know the binary difference operator (makes sense if you think about it).

```python
b - a
```

## Methods for set operations

Fortunately there are some easier to remember methods that return the union, intersection and a few other useful set-theoretic objects. Do notice that these methods don't change `a` or `b` in place:

```python
a.union(b)
```

```python
a
```

```python
b.intersection(c)
```

```python
b
```

```python
{12, 13, 16}.issubset(range(20))
```

```python
set(range(20)).difference(range(10,15))
```

```python
set(range(0, 100, 2)).isdisjoint(range(1, 100, 2))
```

## Cartesian product

You may wonder whether Python has a Cartesian product. It doesn't, but it's pretty easy to replicate using a set comprehension:

```python
a_times_b = {(x, y) for x in a for y in b}
a_times_b
```

At this point you will probably have spotted that all of the above operations (union, intersection, difference) can pretty easily be replicated using a **set comprehension**. I'll leave that as an exercise.
